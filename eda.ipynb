### Признаки запроса

```python
mask = df_train["search_category"] != df_train["item_category_id"]
print(f"Категория запроса не совпадает с категорией выбранного объявления "
      f"в {len(df_train[mask]) / len(df_train) * 100}% случаев")

mask = df_train["search_location_id"] != df_train["item_location_id"]
print(f"Локация запроса не совпадает с локацией выбранного объявления "
      f"в {len(df_train[mask]) / len(df_train) * 100}% случаев")
```

Лишь в 0.01% случаев категория выбранного объявления не совпадает с категорией запроса. Значит для каждой категории запроса будем искать только совпадающие категории объявления.  
Локация запроса совпадает в ~83% случаев. Это много, но не стоит из-за этого делать жёсткий фильтр по локации.

### Цена

```python
print("Максимальная цена объявления:", max(df_train["item_price"]))
print("Минимальная цена объявления: ", min(df_train["item_price"]))
print("Средняя цена объявления: ", df_train["item_price"].mean())

mask = df_train["item_price"] <= 0.0
print(f"Выбранных объявлений с ценой 0 и меньше: {len(df_train[mask]) / len(df_train) * 100}%")
mask = df_train["item_price"] >= 1_000_000.0
print(f"Выбранных объявлений с ценой больше 1_000_000: {len(df_train[mask]) / len(df_train) * 100}%")
print(f"Медиана цены: {df_train["item_price"].median()}")
```

Медиана цены составляет 1000. 7.5% объявлений имеют цену меньше 0 включительно, скорее всего это означает цену по договорённости. Есть выбросы (цена больше миллиардов), но их не так много, только 0.26% объявлений имеют цену больше миллиона.

### Прочие признаки объявления

Я думаю, что стоит убрать признаки item_latitude и item_longitude, так как они явно не могут как-либо охарактеризовать востребуемость объявления.

```python
valid_item_ids = set(df_items["item_id"].unique())
df_train_valid = df_train[df_train["item_id"].isin(valid_item_ids)].copy()
df_train_old = df_train[df_train["item_id"].isin(valid_item_ids) == False].copy()

print(f"Всего строк в train: {len(df_train)}")
print(f"Выбранных объявлений в train, содержащихся в benchmark_items: {len(df_train_valid)}")
print(f"Остальных строк: {len(df_train_old)}")
```

В train содержится 33010/497673 строк, которые присутствуют в benchmark_items. Часть этих данных можно использовать для валидации.
